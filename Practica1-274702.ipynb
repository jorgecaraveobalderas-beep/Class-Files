El codigo abre el archivo de texto "mbox", y recorre todas las lineas del mismo. A su vez, identifica las lineas que unicamente empiecen con "X-DSPAM-Confidence"". Estas lineas siguen siempre la misma estructura. Primero la palabra clave, luego dos puntos ":" y finalmente el valor numerico de la medicion

Despues de esto, considera todo aquello despues de los dos puntos, ya que por el momento, unicamente se buscan los valores de medicion.
Una vez obtenido el valor, al ser cadena, este lo convierte en un valor flotante, ya que los valores extraidos cuentan con decimales.
La conversion de cadena a flotante no es necesario hasta este punto, pero debido a que es posible que estos datos sean utilizados para algun analisis y/o operaciones matematicas.

Para finalizar, se anaden estos datos como valor flotante dentro de un listado, parea que este pueda ser manipulado con mayor facilidad con posibles futuras interacciones con esta informacion.



Calcular y mostrar en pantalla:
La cantidad de datos extraídos.

*   La cantidad de datos extraidos
*   La sumatoria de los datos
*   El promedio de los datos
*   La varianza de los datos

```python
archivo = open('/content/drive/MyDrive/ClassFiles/Clase 1/mbox.txt')
Nueva_Lista = []
for linea in archivo:
  linea = linea.rstrip()
  if linea.startswith('X-DSPAM-Confidence'):
    dato_numerico = linea.find(":")
    mediciones=(linea[dato_numerico+1:])
    mediciones=float(mediciones)
    Nueva_Lista.append(mediciones)

## Cantidad de datos extraidos
Datos_extraidos = len(Nueva_Lista)

## La sumatoria de los datos
Sumatoria_datos = sum(Nueva_Lista)

## El promedio de los datos
Promedio_datos = Sumatoria_datos / Datos_extraidos

## La varianza de los datos
Varianza_datos = sum((x - Promedio_datos) ** 2 for x in Nueva_Lista) / Datos_extraidos



print("La cantidad de datos extraidos es: ", Datos_extraidos)
print("La sumatoria de los datos es: ", Sumatoria_datos)
print("El promedio de los datos es: ", Promedio_datos)
print("La varianza de los datos es: ", Varianza_datos)
```
