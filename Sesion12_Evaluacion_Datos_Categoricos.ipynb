## Actividad 3 — Alta cardinalidad (25 pts)

```python
# 3.1 — Calcula .nunique() para TODAS las columnas del dataset (no solo las categóricas)
# y la razón (valores únicos / total de filas) para cada una.

#Calcular el total de filas

total_filas = len(df)

#Mostrar los valores únicos con la función sugerida .nunique() para todas las columnas y los valores únicos en cada una de estas, seguido calcular el porcentaje del total de valores únicos en cada columna

for columna in df.columns:
    valores_unicos = df[columna].nunique()
    razon = valores_unicos / total_filas
    print(f"{columna:22s} -> {valores_unicos:5d} valores únicos --> ({razon:.1%} del total de filas)")
```

**1.- CustomerID con un 100% de cardinalidad.**

**3.3 — Agrupación "Top 10 + Otros"**

En clase agrupaste `country` de Netflix Titles en sus 10 categorías más frecuentes + `'Otros'`, reduciendo su cardinalidad. Aplica la misma técnica aquí sobre la columna de mayor cardinalidad que identificaste en 3.1 (pista: `.value_counts().head(10)`, luego `.where()` + `.isin()`, igual que en el notebook de clase).

```python
# Aplica el agrupamiento Top 10 + Otros sobre la columna de mayor cardinalidad

# Identificamos las 10 categorías más frecuentes de 'customerID'

top_10 = df['customerID'].value_counts().head(10)

# Guardamos la lista de las categorías que SÍ se conservan
categorias_top10 = top_10.index.tolist()

# Cualquier valor que no esté en el Top 10 se reemplaza por 'Otros'
df['customerID_grouped'] = df['customerID'].where(df['customerID'].isin(categorias_top10), 'Otros')
df['customerID_grouped'].value_counts()
```

_Tu respuesta:_

```python
# 4.2 — Corrige el tipo de TotalCharges.
# Pista: pd.to_numeric() con el parámetro errors= te puede ayudar a identificar
# o manejar los valores problemáticos que encontraste en 4.1.

#Buscar si existen espacios en blanco que causen que se interprete la columna como objeto

espacios_blanco = df[df['TotalCharges'].str.strip() == '']
print(f"Número de filas con espacios en blanco: {len(espacios_blanco)}")

#Al encontrar los espacios en blanco, se convierte a tipo flotante con la funcion sugerida pd.to_numeric() con el parámetro errors=, para convertir los espacios en blanco en Nan, y se comprueba el tipo de variable al final
df['TotalCharges'] = pd.to_numeric(df['TotalCharges'], errors='coerce')
df['TotalCharges'].dtype
```

```python
# 4.3 — Convierte a category las columnas categóricas que, según lo que calculaste
# en la Actividad 3, tengan cardinalidad baja y valores fijos.
# Verifica con .dtypes que el cambio se aplicó correctamente.

#En base al ejercicio anterior, se colocan las columnas con baja cardinalidad

Columnas_baja_cardinalidad = [
    'gender', 'SeniorCitizen', 'Partner', 'Dependents',
    'PhoneService', 'MultipleLines', 'InternetService',
    'OnlineSecurity', 'OnlineBackup', 'DeviceProtection',
    'TechSupport', 'StreamingTV', 'StreamingMovies',
    'Contract', 'PaperlessBilling', 'PaymentMethod', 'Churn'
]

#Utilizar la funcion .astype vista con anterioridad

df[Columnas_baja_cardinalidad] = df[Columnas_baja_cardinalidad].astype('category')

#Utilizar la funcion .dtypes para confirmar la conversion haya sido exitosa

df[Columnas_baja_cardinalidad].dtypes
```
