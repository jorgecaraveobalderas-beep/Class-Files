## Preparación — datasets de trabajo

Usaremos dos datasets a lo largo del notebook: `ventas` y `lecturas`, con datos de sensores — para que practiques los mismos patrones sobre datos distintos.


```python
ventas = [
    {"producto": "Laptop",      "categoria": "Electrónica", "precio": 18500, "unidades": 4},
    {"producto": "Mouse",       "categoria": "Electrónica", "precio": 350,   "unidades": 40},
    {"producto": "Escritorio",  "categoria": "Mobiliario",  "precio": 2200,  "unidades": 12},
    {"producto": "Silla",       "categoria": "Mobiliario",  "precio": 1800,  "unidades": 25},
    {"producto": "Monitor",     "categoria": "Electrónica", "precio": 4200,  "unidades": 9},
    {"producto": "Lámpara",     "categoria": "Mobiliario",  "precio": 600,   "unidades": 18},
]

lecturas = [
    {"sensor": "S1", "hora": 8,  "temperatura": 21.4},
    {"sensor": "S2", "hora": 8,  "temperatura": 19.8},
    {"sensor": "S1", "hora": 12, "temperatura": 27.1},
    {"sensor": "S3", "hora": 12, "temperatura": 25.6},
    {"sensor": "S2", "hora": 16, "temperatura": 23.0},
    {"sensor": "S3", "hora": 16, "temperatura": 24.4},
]

# Lista simple de nombres de producto, construida con un for (todavía sin comprensiones)
productos = []
for v in ventas:
    productos.append(v["producto"])

print(productos)
```

## 1 · `enumerate`

**Recordatorio:** `enumerate(secuencia)` recorre una secuencia y entrega pares `(indice, valor)`, uno por cada elemento. Por defecto el índice empieza en 0; se puede cambiar con el parámetro `start`. Sirve para saber en qué posición está cada elemento sin llevar un contador manual por separado.

### Ejecuta


```python
for indice, nombre in enumerate(productos):
    print(indice, "->", nombre)
```

### Modifica

Cambia `start=0` (el valor por defecto) por `start=100` en la celda de abajo — es una copia de la celda anterior, edítala directamente. Vuelve a correr.

**Antes de correr, responde en un comentario:** ¿qué esperas que cambie? ¿el orden de los productos también cambia?


```python
# TODO — tu predicción:
#

# Modifica esta línea: agrega start=100 dentro de enumerate(...) ## El indice empezara con el valor 100, sin embargo considera los mismo productos
for indice, nombre in enumerate(productos, start = 100):
    print(indice, "->", nombre)
```

### Ejecuta — un uso más realista

`enumerate` es muy útil para construir una "tabla de búsqueda": saber en qué posición está cada elemento.


```python
posicion_producto = {}
for indice, nombre in enumerate(productos):
    posicion_producto[nombre] = indice

print(posicion_producto)
print("El producto 'Silla' está en la posición:", posicion_producto["Silla"])
```

### Modifica

En la celda de abajo, cambia `productos` por `lecturas` y `nombre` por `sensor` (son las dos únicas palabras que necesitas cambiar). Ejecuta y observa: como en `lecturas` cada `sensor` aparece dos veces, ¿qué posición queda guardada al final para cada uno — la primera o la última en la que aparece? ¿Por qué crees que pasa eso?


```python
posicion_sensor = {}
for indice, lectura_item in enumerate(lecturas):
    posicion_sensor[lectura_item["sensor"]] = indice

print(posicion_sensor)

# Tu observación:
# La última posición en la que aparece el sensor es la que queda guardada, porque el diccionario se sobrescribe.
# Por ejemplo, para 'S1', queda la posición 2 (la última vez que aparece en 'lecturas').
```

## 2 · `sorted`

**Recordatorio:** `sorted(secuencia)` devuelve una lista nueva y ordenada; nunca modifica la secuencia original. Cuando la secuencia contiene diccionarios, `sorted` no sabe por cuál campo comparar — se lo decimos con el parámetro `key`. `operator.itemgetter("campo")` crea justo eso: un extractor que, dado un diccionario, regresa el valor de ese campo. El parámetro `reverse=True` invierte el orden.

Este es el bloque más largo de la actividad — tómate tu tiempo, especialmente en la parte de múltiples criterios.

### Ejecuta


```python
numeros = [7, 1, 2, 6, 0, 3, 2]
print(sorted(numeros))
print(sorted(numeros, reverse=True))
```

### Modifica

En la celda de abajo, cambia la lista `numeros` por una con decimales o negativos (es una copia editable). Vuelve a correr: ¿`sorted()` se comporta distinto?


```python
# TODO — modifica esta lista (agrega decimales o negativos)
numeros_copia = [7, 1, 2, 6, 0, 3, 2,-2,33.2,1.2,-1000]
print(sorted(numeros_copia))
```

### Ejecuta — ordenar diccionarios con `operator.itemgetter`


```python
orden_por_precio = sorted(ventas, key=operator.itemgetter("precio"), reverse=False)

for v in orden_por_precio:
    print(v["producto"], "-", v["precio"])
```

### Predicción

Sin correr la celda todavía: si cambias `reverse=True` por `reverse=False`, ¿qué producto queda primero? Escríbelo en el comentario, luego corre y verifica.


```python
# TODO — tu predicción: el primero será _Mouse__________

orden_por_precio_asc = sorted(ventas, key=operator.itemgetter("precio"), reverse=False)

for v in orden_por_precio_asc:
    print(v["producto"], "-", v["precio"])
```

### Modifica

En la celda de abajo, cambia `ventas` por `lecturas` y `"precio"` por `"temperatura"` (son las dos cosas que necesitas cambiar) para ordenar `lecturas` de mayor a menor temperatura.


```python
# TODO — cambia ventas por lecturas, y "precio" por "temperatura"
orden_variante = sorted(lecturas, key=operator.itemgetter("temperatura"), reverse=True)

for v in orden_variante:
    print(v)
```

### Ejecuta — múltiples criterios, misma dirección

`itemgetter` acepta más de un campo a la vez: `operator.itemgetter("categoria", "precio")` regresa una tupla `(categoria, precio)`, y `sorted` compara esas tuplas campo por campo. Esto funciona bien cuando **ambos** campos van en la misma dirección (los dos ascendentes, por ejemplo).


```python
orden_multi = sorted(ventas, key=operator.itemgetter("categoria", "precio"), reverse=True)

for v in orden_multi:
    print(v["categoria"], "-", v["producto"], "-", v["precio"])
```

### Ejecuta — múltiples criterios, direcciones distintas

¿Y si quieres categoría ascendente **pero** precio descendente dentro de cada categoría? `itemgetter` no permite mezclar direcciones en una sola llamada. La solución usa algo que ya viste: `sorted` es estable. Si ordenas primero por el criterio secundario y después por el principal, el segundo ordenamiento respeta el orden que dejó el primero para los empates.


```python
# Paso 1: ordenar por precio, de mayor a menor (criterio secundario)
paso1 = sorted(ventas, key=operator.itemgetter("precio"), reverse=True)

# Paso 2: ordenar ese resultado por categoría, ascendente (criterio principal)
# Como sorted es estable, dentro de cada categoría se conserva el orden por precio del paso 1
orden_multi_mixto = sorted(paso1, key=operator.itemgetter("categoria"))

for v in orden_multi_mixto:
    print(v["categoria"], "-", v["producto"], "-", v["precio"])
```

### Modifica

En la celda de abajo (una copia de la técnica de dos pasos), cambia `ventas` por `lecturas`, `"precio"` por `"temperatura"` y `"categoria"` por `"sensor"` — para ordenar `lecturas` primero por `temperatura` de mayor a menor, y después por `sensor` ascendente.


```python
# TODO — adapta esta celda a lecturas / temperatura / sensor

orden_variante = sorted(lecturas, key=operator.itemgetter("sensor"), reverse=True)
paso1_variante = sorted(orden_variante, key=operator.itemgetter("temperatura"), reverse=True)


for v in paso1_variante:
    print(v)

# Pregunta para pensar: ¿por qué el orden de los dos pasos importa?
# (pista: intenta invertir el orden de los dos sorted, y compara)
```

## 3 · `zip`

**Recordatorio:** `zip(secuencia1, secuencia2, ...)` empareja los elementos de varias secuencias según su posición, generando tuplas. Acepta dos o más secuencias a la vez. Si las secuencias no tienen la misma longitud, `zip` se detiene en la más corta, sin mostrar ningún error ni advertencia.

### Ejecuta


```python
sensores = []
temperaturas = []
for l in lecturas:
    sensores.append(l["sensor"])
    temperaturas.append(l["temperatura"])

pares = list(zip(sensores, temperaturas))
print(pares)
```

### Modifica

En la celda de abajo, agrega una tercera lista (`horas`, constrúyela con un `for` igual que `sensores` y `temperaturas`) y agrégala al `zip`. Observa que ahora cada tupla tiene tres elementos, no dos.


```python
# TODO — agrega la lista horas y súmala al zip

horas = []
sensores = []
temperaturas = []

for l in lecturas:
    sensores.append(l["sensor"])
    temperaturas.append(l["temperatura"])
    horas.append(l["hora"])

pares = list(zip(sensores, temperaturas,horas))
print(pares)
```

### Ejecuta — el patrón `zip` + `enumerate`


```python
for indice, (sensor, temperatura) in enumerate(zip(sensores, temperaturas), start=1):
    print(f"Lectura {indice}: {sensor} registró {temperatura}°C")
```

### Modifica

En la celda de abajo, recorta una de las dos listas antes del `zip` (por ejemplo, cambia `temperaturas` por `temperaturas[:-2]` dentro del `zip`) para provocar a propósito el truncamiento silencioso. Cuenta cuántas tuplas produce el resultado.


```python
# TODO — recorta una de las dos listas dentro del zip (por ejemplo, temperaturas[:-2])
resultado_variante = list(zip(sensores, temperaturas[:-2]))
print(resultado_variante)
print("Tuplas producidas:", len(resultado_variante))
```

## 4 · `reversed`

**Recordatorio:** `reversed(secuencia)` recorre una secuencia de atrás hacia adelante, sin modificar la original. El resultado no es una lista, sino un objeto que se genera bajo demanda; para verlo completo o guardarlo se envuelve con `list(reversed(secuencia))`.

### Ejecuta


```python
horas_unicas = [8, 12, 16]

for h in reversed(horas_unicas):
    print(h)
```

### Modifica

En la celda de abajo, cambia `horas_unicas` por `productos`. Guarda el resultado en una lista con `list(...)` y confirma que `productos` no cambió.


```python
# TODO — cambia horas_unicas por productos

resultado_variante = list(reversed(productos))
print("resultado_variante:", resultado_variante)
print("productos (sin cambios):", productos)
```

### Predicción

¿Qué crees que imprime este código? Escribe tu predicción antes de correr — presta atención a qué número acompaña a cada sensor.


```python
# TODO — tu predicción:
#

for indice, sensor in enumerate(reversed(sensores)):
    print(indice, "->", sensor)
```

**Pista si te sorprendió el resultado:** `enumerate` numera la secuencia que recibe. Si le das una secuencia ya invertida, numera *esa*, empezando otra vez desde 0 — no "recuerda" las posiciones originales.
