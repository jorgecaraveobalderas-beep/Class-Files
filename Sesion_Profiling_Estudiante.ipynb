## Parte 1 — Funciones a nivel Dataset

Cada función recibe el DataFrame completo (`df`) y regresa **un solo valor**. Completa el cuerpo de cada función donde dice `# TODO`.

```python
def n_columnas(df):
    return df.shape[1]
    pass
print ("Numero de columnas",n_columnas(df))


def n_filas(df):
    return df.shape[0]
    pass
print ("Numero de filas",n_filas(df))

```

```python
def n_datos_ausentes(df):
  return df.isna().sum().sum()
  pass
print ("Numero de datos ausentes",n_datos_ausentes(df))
    # """Regresa el número total de celdas con datos ausentes (NaN) en todo el DataFrame."""



def pct_celdas_ausentes(df):
  return (df.isna().sum().sum() / df.size) * 100
  pass
print ("Porcentaje de datos ausentes",pct_celdas_ausentes(df))

    #  """Regresa el porcentaje de celdas con datos ausentes sobre el total de celdas
   #  (filas x columnas) del DataFrame."""
    # TODO
```

```python
def n_filas_duplicadas(df):
    """Regresa el número de filas duplicadas en el DataFrame."""
    return df.duplicated().sum()
    # TODO
    pass

print ("Numero de filas duplicadas",n_filas_duplicadas(df))

def pct_filas_duplicadas(df):
    """Regresa el porcentaje de filas duplicadas sobre el total de filas."""
    return (df.duplicated().sum() / df.shape[0]) * 100
    # TODO
    pass


print ("Porcentaje de filas duplicadas",pct_filas_duplicadas(df))
```

```python
def n_variables_numericas(df):
    """Regresa el número de columnas con tipo de dato numérico (int o float)."""
    return df.select_dtypes(include=['int64',"float64"]).shape[1]
    # TODO
    pass

print ("Numero de variables numericas",n_variables_numericas(df))

def n_variables_booleanas(df):
    """Regresa el número de columnas con tipo de dato booleano."""
    return df.select_dtypes(include=['bool']).shape[1]
      # TODO
    pass


print ("Numero de variables booleanas",n_variables_booleanas(df))

def n_variables_categoricas(df):
    """Regresa el número de columnas con tipo de dato object o category."""
    return df.select_dtypes(include=['object','category']).shape[1]
    # TODO
    pass

print ("Numero de variables categoricas",n_variables_categoricas(df))
```

## Parte 2 — Funciones a nivel Columna

Cada función recibe el DataFrame y el **nombre de una columna** (`col`), y regresa un solo valor para esa columna.

```python
col = df.columns

def total_ausentes(df, col):
  return df[col].isna().sum()
     # """Regresa el número de valores ausentes (NaN) en la columna col."""
    # TODO


print ("Numero de valores ausentes",total_ausentes(df,col))


def pct_ausentes(df, col):
  return (df[col].isna().sum() / df.shape[0]) * 100
  pass
print ("Porcentaje de valores ausentes",pct_ausentes(df,col))
  #   """Regresa el porcentaje de valores ausentes en la columna col."""
    # TODO
```

```python

col_new = df.select_dtypes(include=['int64',"float64"]).columns


def media(df, col_new):
  return df[col_new].mean()
  pass
print ("Media",media(df,col_new))
    # """Regresa la media de la columna col (asume que es numérica)."""
    # TODO


def minimo(df, col_new):
  return df[col_new].min()
  pass
print ("Minimo",minimo(df,col_new))

    # """Regresa el valor mínimo de la columna col."""
    # TODO


def maximo(df, col_new):
  return df[col_new].max()
  pass
print ("Maximo",maximo(df,col_new))
    # """Regresa el valor máximo de la columna col."""
    # TODO
```

```python
def total_ceros(df, col):
  return (df[col] == 0).sum()
  pass
print ("Numero de valores iguales a 0",total_ceros(df,col))

   # """Regresa el número de valores iguales a 0 en la columna col."""
    # TODO


def pct_ceros(df, col):
  return (df[col] == 0).sum() / df.shape[0] * 100
  pass
print ("Porcentaje de valores iguales a 0",pct_ceros(df,col))
   ##  """Regresa el porcentaje de valores iguales a 0 en la columna col."""
    # TODO
```

```python
def total_negativos(df, col_new):
  return (df[col_new] < 0).sum()
  pass
print ("Numero de valores negativos",total_negativos(df,col_new))
   ##  """Regresa el número de valores negativos en la columna col."""
    # TODO


def pct_negativos(df, col_new):
  return (df[col_new] < 0).sum() / df.shape[0] * 100
  pass
print ("Porcentaje de valores negativos",pct_negativos(df,col_new))
    ## """Regresa el porcentaje de valores negativos en la columna col."""
    # TODO
```
