## Actividad 1. Conversión y validación (20 puntos)

### 1.1 Conversión (5 pts)

1. Muestra el tipo de dato de `pickup` y `dropoff`.
2. Convierte ambas columnas a tipo fecha con `pd.to_datetime()`, indicando el **formato explícito** que corresponde a los valores.
3. Verifica que la conversión no generó `NaT`: compara el número de faltantes antes y después.

```python
#1


Tipo_de_dato_pickup = taxis['pickup'].dtype
Tipo_de_dato_dropoff = taxis['dropoff'].dtype

print("El tipo de dato para pickup es,", Tipo_de_dato_pickup)
print("El tipo de dato para dropoff es,", Tipo_de_dato_dropoff)

valores_faltantes_orignal = (taxis[['pickup', 'dropoff']].isna().sum())
print(valores_faltantes_orignal)

#Al confirmar que son tipo string, se realiza la conversion a tipo fecha

#2

taxis['pickup'] = pd.to_datetime(taxis['pickup'], errors= 'coerce', format='%Y-%m-%d %H:%M:%S') #El formato en el dataset es 'YYYY-MM-DD HH:MM:SS
taxis['dropoff'] = pd.to_datetime(taxis['dropoff'], errors='coerce', format='%Y-%m-%d %H:%M:%S') #El formato en el dataset es 'YYYY-MM-DD HH:MM:SS


Nuevo_tipo_de_dato_pickup = taxis['pickup'].dtype
Nuevo_tipo_de_dato_dropoff = taxis['dropoff'].dtype

print("El nuevo tipo de dato para pickup es,", Nuevo_tipo_de_dato_pickup)
print("El nuevo tipo de dato para dropoff es,", Nuevo_tipo_de_dato_dropoff)

#3

valores_faltantes_nuevo = (taxis[['pickup', 'dropoff']].isna().sum())
print(valores_faltantes_nuevo)

#Durante la conversion no genero valores NaT, lo cual quiere decir que en ambas columnas todos los registros mantienen el mismo formato.
```

### 1.2 Periodo de los datos (5 pts)

### 1.3 Cambio de horario (5 pts)

El 10 de marzo de 2019, en Nueva York, las horas entre 02:00 y 02:59 **no existieron**.

1. Filtra los viajes con `pickup` el 10 de marzo y cuenta cuántos hay en cada hora (`.dt.hour` y `groupby()` o `value_counts()`). ¿Hay viajes a las 2:00?
2. Localiza `pickup` en la zona `'America/New_York'` con `tz_localize()` usando `nonexistent='NaT'` y cuenta cuántos valores quedan como `NaT`.

```python
#1

marzo_10_2019 = taxis[taxis['pickup'].dt.date == pd.to_datetime('2019-03-10').date()]
conteo_horas = marzo_10_2019['pickup'].dt.hour.value_counts().sort_index() # Contar la frecuencia de viajes por cada hora del día y ordenar el resultado cronológicamente
print(conteo_horas)

# Crear máscara booleana para verificar específicamente si existen viajes iniciados en la hora 2 (2:00 a 2:59)

viajes_a_las_2 = (marzo_10_2019['pickup'].dt.hour == 2)
print(f"Número de viajes a las 2:00: {len(marzo_10_2019[viajes_a_las_2])}")

#No hay viajes a las 2:00

#2

pick_up_zona = taxis['pickup'].dt.tz_localize('America/New_York', nonexistent='NaT') #Localizar la zona horaria de NY asignando NaT a las horas inexistentes por el cambio de horario de verano
conteo_nat = (pick_up_zona.isna().sum())
print (f"Cantidad de valores que quedaron como NaT: {conteo_nat}")
```

## Actividad 2. Duraciones y anomalías (25 puntos)

### 2.1 Duración de cada viaje (6 pts)

1. Crea la columna `duracion` = `dropoff` − `pickup`.
2. Crea `duracion_min` con la duración en **minutos** (usa `.dt.total_seconds()`).
3. Muestra el resumen estadístico de `duracion_min` con `describe()`.

```python
#1

taxis['duracion'] = taxis['dropoff'] - taxis['pickup']

#2
taxis['duracion_min'] = taxis['duracion'].dt.total_seconds() / 60 #Convierte el timedelta a segundos totales y lo divide entre 60 para obtener el tiempo en minutos

#3

resumen_estadistico = taxis['duracion_min'].describe() #Genera y muestra el resumen estadístico descriptivo (conteo, media, min, max, etc.) de la duración en minutos
print(resumen_estadistico)
```

Muestra los viajes de duración cero con las columnas `pickup`, `dropoff`, `distance`, `fare` y `total`.

### 2.3 Velocidad promedio (6 pts)

La velocidad promedio de un viaje, en millas por hora, es `distance / (duracion_min / 60)`.

1. Crea la columna `velocidad_mph` **solo para viajes con duración mayor que cero**; en los demás debe quedar como faltante (`NaN`). Sugerencia: calcula la velocidad para todos los viajes y conserva solo los valores válidos con `.where(condición)`, que deja `NaN` donde la condición es `False`.
2. Cuenta los viajes con velocidad mayor a 60 mph (velocidad poco plausible en tráfico urbano).
3. Muestra los 5 viajes con mayor velocidad, ordenados de mayor a menor, con `pickup`, `dropoff`, `duracion_min`, `distance`, `velocidad_mph` y `fare`.

```python
#1

velocidad_mph = taxis['distance'] / (taxis['duracion_min'] / 60) #Calcula la velocidad en mph e asigna NaN a los viajes con duración menor o igual a cero usando .where()
taxis['velocidad_mph'] = velocidad_mph.where(taxis['duracion_min'] > 0)
print(taxis[taxis['velocidad_mph'].isna()]['velocidad_mph'])

#2

velocidad_mayor_60 = (taxis['velocidad_mph'] > 60).sum() #Cuenta el total de viajes cuya velocidad es superior a 60 mph
print(f"Viajes con velocidad mayor a 60 mph: {velocidad_mayor_60}")

#3

viajes_velocidad_mayor = taxis.sort_values(by='velocidad_mph', ascending=False).head(5) #Ordena los viajes descendentemente por velocidad y muestra los primeros 5 con las columnas requeridas
display(viajes_velocidad_mayor[['pickup', 'dropoff', 'duracion_min', 'distance', 'velocidad_mph', 'fare']])
```

## Actividad 3. Componentes y patrones (25 puntos)

### 3.1 Extraer componentes (6 pts)

Crea las columnas:

- `hora`: hora de `pickup`.
- `dia_num`: día de la semana de `pickup` (lunes = 0).
- `dia_nombre`: nombre del día de `pickup`.
- `fin_semana`: `True` si es sábado o domingo.
- `fecha`: solo la fecha (sin hora) de `pickup`.
- `cruza_medianoche`: `True` si la **fecha** de `dropoff` es distinta de la de `pickup`.

Cuenta cuántos viajes cruzan la medianoche.

```python
taxis['hora'] = taxis['pickup'].dt.hour #Extrae la hora en formato de entero (0-23) de la fecha de pickup
taxis['dia_num'] = taxis['pickup'].dt.dayofweek #Extrae el número del día de la semana (0=Lunes,1=Martes...)
taxis['dia_nombre'] = taxis['pickup'].dt.day_name() #Extrae el nombre en texto del día de la semana
taxis['fin_semana'] = taxis['dia_num'].isin([5, 6]) #Asigna True si el día es sábado (5) o domingo (6), de lo contrario False
taxis['fecha'] = taxis['pickup'].dt.date #Extrae únicamente la fecha (año-mes-día)
taxis['cruza_medianoche'] = taxis['dropoff'].dt.date != taxis['pickup'].dt.date #Compara si la fecha de dropoff es distinta a la de pickup para identificar si cruzó la medianoche


columnas_nuevas = ['hora', 'dia_num', 'dia_nombre', 'fin_semana', 'fecha', 'cruza_medianoche']
display(taxis[columnas_nuevas].head(5))

viajes_medianoche = taxis['cruza_medianoche'].sum()
print(f"Viajes que cruzan la medianoche: {viajes_medianoche}")
```

### 3.2 Patrón por hora del día (7 pts)

Con `groupby('hora')`, calcula para cada hora: el **número de viajes** y la **mediana** de `duracion_min`. Indica la hora con más viajes y la hora con menos.

Usa la mediana (no la media) porque la duración tiene valores extremos.

```python
group_por_hora = taxis.groupby('hora') #Agrupa el DataFrame por la columna 'hora'
numero_viajes_por_hora = group_por_hora.size() #Obtiene la cantidad total de viajes por cada hora
mediana_duracion_por_hora = group_por_hora['duracion_min'].median() #Calcula la mediana de la duración en minutos para cada hora

hora_max = numero_viajes_por_hora.idxmax()
val_max = numero_viajes_por_hora.max()

hora_min = numero_viajes_por_hora.idxmin()
val_min = numero_viajes_por_hora.min()

print(f"La hora con más viajes es {hora_max} con {val_max} viajes.")
print(f"La hora con menos viajes es {hora_min} con {val_min} viajes.")

print("Número de viajes por hora:")
print(numero_viajes_por_hora)
print("\nMediana de duración por hora:")
print(mediana_duracion_por_hora)
```

### 3.3 Patrón por día de la semana (6 pts)

Calcula por día de la semana el **número de viajes** y la **mediana** de `duracion_min` y de `fare`. Presenta los días en orden de lunes a domingo (sugerencia: agrupa por `['dia_num', 'dia_nombre']`).

Toma en cuenta que marzo de 2019 no tiene el mismo número de cada día de la semana: calcula también el **promedio de viajes por fecha**, dividiendo entre el número de fechas distintas de ese día de la semana (usa la columna `fecha` y `.nunique()`).

```python
patron_dias = taxis.groupby(['dia_num', 'dia_nombre']) #Agrupa por el número y nombre del día para mantener de forma natural el orden de lunes a domingo
num_viajes = patron_dias['duracion_min'].count()
mediana_duracion = patron_dias['duracion_min'].median()
mediana_fare = patron_dias['fare'].median()
dias_distintos = patron_dias['fecha'].nunique()

patron_dias = pd.DataFrame({
    'num_viajes': num_viajes,
    'dias_distintos': dias_distintos,
    'mediana_duracion': mediana_duracion,
    'mediana_fare': mediana_fare
}).reset_index() #Consolida todas las métricas en un nuevo DataFrame (reinicia el index a 0)

patron_dias['promedio_viajes_por_dia'] = patron_dias['num_viajes'] / patron_dias['dias_distintos']
columnas_orden = ['dia_nombre', 'num_viajes', 'dias_distintos', 'promedio_viajes_por_dia', 'mediana_duracion', 'mediana_fare']
print(patron_dias[columnas_orden])
```
