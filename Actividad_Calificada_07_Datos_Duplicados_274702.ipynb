### 1.1 Conteo de duplicados exactos (5 pts)

Calcula el número de filas marcadas como duplicadas en `hoteles` considerando **todas** las columnas, con `keep='first'`, `keep='last'` y `keep=False`. Imprime los tres resultados con un texto que explique qué significa cada uno.

```python

#Calcular las filas duplicadas del DataFrame "hoteles" usando keep="first", "last" y "False"

Filas_duplicadas_first = hoteles.duplicated(keep='first').sum()
Filas_duplicadas_last = hoteles.duplicated(keep='last').sum()
Filas_duplicadas_false = hoteles.duplicated(keep=False).sum()

print(f"Numero de filas duplicadas (keep='first'): {Filas_duplicadas_first}")
print("Indica cuantas filas son repeticiones conservando unicamente la primer observacion")

print(f"\nNumero de filas duplicadas (keep='last'): {Filas_duplicadas_last}")
print("Indica cuantas filas son repeticiones conservando unicamente la ultima observacion")


print(f"\nNumero de filas duplicadas (keep=False): {Filas_duplicadas_false}")
print("Indica cuantas filas son parte de algun grupo duplicado")
```

Con los resultados de 1.1, calcula e imprime:

- El **porcentaje** de filas sobrantes respecto al total.

```python
#1

total_filas = len(hoteles)
porcentaje_filas_sobrantes = (Filas_duplicadas_first / total_filas) * 100

print(f"El porcentaje de filas sobrantes respecto al total es: {porcentaje_filas_sobrantes:.2f}%") #Calcular el porcentaje de filas repetidas

#2

grupos_repetidos = Filas_duplicadas_false - Filas_duplicadas_first #Calcular número de combinaciones de valores que aparecen más de una vez

print(f"El numero de grupos repetidos es: {grupos_repetidos}")
```

### 1.3 Duplicados por hotel (5 pts)

Determina **cuántas** filas sobrantes (`keep='first'`) hay en cada hotel y qué **porcentaje** representan respecto al total de reservas de ese hotel.

Sugerencia: crea una copia de `hoteles`, agrega una columna booleana `es_duplicado` con el resultado de `duplicated()` y agrupa por `hotel`.

```python
df_hoteles = hoteles.copy() #Crear una copia del dataframe original

df_hoteles['es_duplicado'] = df_hoteles.duplicated(keep='first') #Filas duplicadas considerando el primer registro

duplicados_por_hotel = df_hoteles.groupby('hotel')['es_duplicado'].sum()
duplicados_por_hotel_porcentaje = (duplicados_por_hotel / hoteles.groupby('hotel').size()) * 100

print(duplicados_por_hotel)
print(duplicados_por_hotel_porcentaje)
```

## Actividad 3. Impacto en el análisis (25 puntos)

### 3.1 Versión sin duplicados (5 pts)

Crea `hoteles_limpio` eliminando los duplicados exactos, conservando la primera aparición y **renumerando el índice**. Imprime el número de filas antes y después y cuántas se eliminaron.

```python
hoteles_limpio = hoteles.drop_duplicates(keep='first').reset_index(drop=True)

filas_eliminadas = len(hoteles) - len(hoteles_limpio)

print(f"Filas antes: {len(hoteles)}") #Total de filas dataframe original
print(f"Filas despues: {len(hoteles_limpio)}") #Total filas sin duplicados
print(f"Filas_eliminas: {filas_eliminadas}") #Delta original vs sin duplicados
```

### 3.2 Métricas por hotel (8 pts)

Para `hoteles` y para `hoteles_limpio`, calcula por hotel con `groupby()`:

- Número de reservas (`.size()`).
- Tasa de cancelación (media de `is_canceled`).
- Tarifa diaria promedio (media de `adr`).

Nota: para calcular la media de varias columnas a la vez, selecciona una lista de columnas: `df.groupby('hotel')[['col1', 'col2']].mean()`.

Guarda las medias en dos `DataFrame`, `resumen_original` y `resumen_limpio`, con `hotel` como columna (usa `reset_index()`).

```python
resumen_original = hoteles.groupby('hotel')[['is_canceled', 'adr']].mean().reset_index() #Promedio de cancelación y ADR por hotel
resumen_original['reservas'] = hoteles.groupby('hotel').size().values #Conteo total de reservas por hotel
resumen_original = resumen_original.rename(columns={'is_canceled': 'is_canceled_original', 'adr': 'adr_original'})

resumen_limpio = hoteles_limpio.groupby('hotel')[['is_canceled', 'adr']].mean().reset_index()  #Promedio de cancelación y ADR por hotel sin duplicados
resumen_limpio['reservas'] = hoteles_limpio.groupby('hotel').size().values #Conteo total de reservas por hotel sin duplicados
resumen_limpio = resumen_limpio.rename(columns={'is_canceled': 'is_canceled_sin_duplicados', 'adr': 'adr_sin_duplicados'})

display(resumen_original)
display(resumen_limpio)

```

### 3.3 Tabla comparativa (6 pts)

Agrega a cada resumen una columna `version` (`'original'` y `'sin duplicados'`) y únelos con `pd.concat()` en un `DataFrame` llamado `comparacion`, con índice renumerado y ordenado por `hotel`, de modo que las dos versiones de cada hotel queden juntas.

```python
resumen_original = resumen_original.copy()
resumen_original = resumen_original.rename(columns={'is_canceled_original': 'is_canceled', 'adr_original': 'adr'}) #Estandariza los nombres de las columnas antes de concatenar para evitar valores "NaN" ya que identifica que son columnas distintas
resumen_original['version'] = 'original'

resumen_limpio = resumen_limpio.copy()
resumen_limpio = resumen_limpio.rename(columns={'is_canceled_sin_duplicados': 'is_canceled', 'adr_sin_duplicados': 'adr'}) #Estandariza los nombres de las columnas antes de concatenar para evitar valores "NaN" ya que identifica que son columnas distintas
resumen_limpio['version'] = 'sin duplicados'

comparacion = pd.concat([resumen_original, resumen_limpio], ignore_index=True) #Se unen ambos dataframes y se considera el index desde 0 por columna para evitar duplicados
comparacion = comparacion.sort_values(by=['hotel', 'version']).reset_index(drop=True) #Se sortea y se considera el index desde 0 por fila para evitar duplicados
display(comparacion)
```
