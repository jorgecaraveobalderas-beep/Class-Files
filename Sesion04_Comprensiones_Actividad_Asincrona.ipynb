## 1 · Comprensión de listas

**Recordatorio:** una comprensión de listas crea una lista nueva en una sola línea, con la forma `[expresión for elemento in secuencia]`, opcionalmente con un `if` al final para filtrar. Es exactamente el mismo patrón que `lista = []` + `for` + `.append(...)`, escrito de forma compacta.

### Ejecuta


```python
# La forma larga que ya conoces:
productos_for = []
for v in ventas:
    productos_for.append(v["producto"])

# La misma idea, como comprensión:
productos_comp = [v["producto"] for v in ventas]

print(productos_for)
print(productos_comp)
print(productos_for == productos_comp)
```

### Predicción

Sin correr la celda: ¿qué tipo de dato regresa `[v["precio"] for v in ventas]` — una lista, un diccionario, o un solo número? Escribe tu predicción, luego corre y verifica.


```python
# TODO — tu predicción:
# LISTA

precios = [v["precio"] for v in ventas]
print(type(precios))
print(precios)
```

### Ejecuta — con condición


```python
caros = [v["producto"] for v in ventas if v["precio"] > 2000]
print(caros)
```

### Modifica

En la celda de abajo (una copia editable), cambia el número `2000` por otro valor, y también cambia `v["precio"]` por `v["unidades"]` para filtrar por otro campo. Observa cómo cambia el resultado.

```python
ingresos = [v["precio"] * v["unidades"] for v in ventas]
print(ingresos)
```

## 2 · Comprensión de diccionarios

**Recordatorio:** una comprensión de diccionarios crea un diccionario nuevo en una sola línea, con la forma `{llave: valor for elemento in secuencia}` — el mismo patrón que `dict = {}` + `for` + asignación por llave, escrito de forma compacta. También acepta `if` al final para filtrar.

### Ejecuta


```python
# La forma larga que ya conoces:
precio_por_producto_for = {}
for v in ventas:
    precio_por_producto_for[v["producto"]] = v["precio"]

# La misma idea, como comprensión:
precio_por_producto_comp = {v["producto"]: v["precio"] for v in ventas}

precio_por_producto_comp = {a["producto"] : a["precio"] for a in ventas}


print(precio_por_producto_for)
print(precio_por_producto_comp)
print(precio_por_producto_for == precio_por_producto_comp)
```

### Modifica

En la celda de abajo, cambia la llave de `v["producto"]` a `v["categoria"]`. Ejecuta y observa: como `categoria` solo tiene dos valores posibles y se repite entre varios productos, ¿qué pasa con el diccionario resultante? ¿Cuántas llaves termina teniendo?


```python
# TODO — cambia v["producto"] por v["categoria"] como llave
resultado_variante = {v["categoria"]: v["precio"] for v in ventas}
print(resultado_variante)

# Tu observación:
#
```

### Ejecuta — con condición


```python
precios_caros = {
    v["producto"]: v["precio"]
    for v in ventas
    if v["precio"] > 2000
}
print("Los precios mas caros que tenemos son", precios_caros)
```

### Predicción

Sin correr la celda: ¿cuántas llaves tendrá el diccionario que arma el siguiente código, dado que `lecturas` tiene 6 registros pero solo 3 sensores distintos (cada uno repetido dos veces)? Escribe tu predicción, luego corre y verifica.


```python
# TODO — tu predicción: el diccionario tendrá __3__ llaves

temperatura_por_sensor = {l["sensor"]: l["temperatura"] for l in lecturas}
print(temperatura_por_sensor)
print("Número de llaves:", len(temperatura_por_sensor))
```

**Pista si tu predicción no coincidió:** en un diccionario no puede haber llaves repetidas. Si dos elementos generan la misma llave, el segundo sobrescribe al primero — por eso el resultado tiene menos llaves que elementos tenía `lecturas`. Es el mismo comportamiento que viste en la actividad anterior con `enumerate` y llaves repetidas.

## 3 · Comprensión de conjuntos


**Recordatorio:** una comprensión de conjuntos usa `{ }`, igual que la de diccionarios, pero con una sola expresión — `{expresión for elemento in secuencia}` — así que crea un **set**, no un diccionario. Un set nunca tiene elementos repetidos y no garantiza un orden. Es la comprensión de listas ideal cuando lo que te interesa son los valores *únicos*, no la lista completa.

### Ejecuta


```python
# La forma larga que ya conoces:
categorias_for = set()
for v in ventas:
    categorias_for.add(v["categoria"])

# La misma idea, como comprensión:
categorias_comp = {v["categoria"] for v in ventas}

print(categorias_for)
print(categorias_comp)
print(type(categorias_comp))
```

### Predicción

`ventas` tiene 6 registros, pero `categoria` solo toma dos valores posibles ("Electrónica" o "Mobiliario"). Antes de correr: ¿cuántos elementos tendrá `{v["categoria"] for v in ventas}`? Escribe tu predicción, luego corre y verifica.


```python
# TODO — tu predicción: el conjunto tendrá __2__ elementos

categorias = {v["categoria"] for v in ventas}
print(categorias)
print("Número de elementos:", len(categorias))
```

## 4 · Comprensiones anidadas — *flattening*


**Recordatorio:** una comprensión anidada escribe varios `for`, uno detrás de otro, dentro de los mismos corchetes — en el mismo orden en que los escribirías como bloques `for` separados. El uso más común es "aplanar" (*flatten*) una lista de listas en una sola lista plana.

### Ejecuta — el problema: una lista de listas


```python
lecturas_por_sensor = [
    [21.4, 27.1],  # temperaturas del sensor S1
    [19.8, 23.0],  # temperaturas del sensor S2
    [25.6, 24.4],  # temperaturas del sensor S3
]

print(lecturas_por_sensor)
```

### Ejecuta — la forma larga: `for` dentro de `for`


```python
todas_las_temperaturas_for = []
for sensor in lecturas_por_sensor:
    for temp in sensor:
        todas_las_temperaturas_for.append(temp)

print(todas_las_temperaturas_for)
```

### Ejecuta — la misma idea, como comprensión anidada

Fíjate en el orden: la expresión va primero, después el `for` exterior, después el `for` interior — exactamente el mismo orden en que aparecen los bloques arriba.


```python
todas_las_temperaturas_comp = [
    temp
    for sensor in lecturas_por_sensor
    for temp in sensor
]

print(todas_las_temperaturas_comp)
print(todas_las_temperaturas_for == todas_las_temperaturas_comp)
```

### Predicción

Si agregas `if temp > 23` al final de la comprensión anidada de arriba, ¿cuántos elementos esperas que queden? Escribe tu predicción, luego corre y verifica.


```python
# TODO — tu predicción: quedarán __3__ elementos

temperaturas_altas = [
    temp
    for sensor in lecturas_por_sensor
    for temp in sensor
    if temp > 23
]
print(temperaturas_altas)
```

### Modifica

En la celda de abajo, cambia el valor `23` por otro número, y observa cómo cambia cuántos elementos quedan.


```python
# TODO — modifica el valor de comparación
resultado_variante = [
    temp
    for sensor in lecturas_por_sensor
    for temp in sensor
    if temp > 26
]
print(resultado_variante)
```

**Recordatorio importante:** si una comprensión anidada necesita más de dos `for`, o mezclas varios `if`, deja de ser más clara que el `for` tradicional — en ese caso, vuelve al `for` con bloques. Anidar no es la meta; la claridad sí.
